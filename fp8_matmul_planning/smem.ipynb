```python
bk_stages = 2
c_stages = 2

def get_smem_kilobytes(BM,BN,BK,bk_stages,c_stages): 
  As_bytes = BM*BK 
  Bs_bytes = BK*BN 
  Cs_bytes = 2*BM*BN
  total_bytes = (bk_stages*(As_bytes+Bs_bytes)) + (c_stages*(Cs_bytes))
  return total_bytes/(1000)
```

```python
BM,BK,BN = 64,64,64

total_bytes = get_smem_kilobytes(BM,BN,BK,bk_stages,c_stages)

```
