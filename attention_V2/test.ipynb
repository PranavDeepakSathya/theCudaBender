```python
import torch
```

```python
LQ = 8192
D = 128 

Q = torch.randn(LQ,D, dtype=torch.bfloat16, device = "cuda")
```

```python
Q_r = Q.reshape(2*LQ, D//2)
```

```python
x = 32 
y = 17 

Q[x,y] == Q_r[(2*x) + (y//(D//2)), y % (D//2)]


Q_stride = Q.stride()
Q_r_stride = Q_r.stride()
```

```python
Q.flatten()[x*Q_stride[0] + y*Q_stride[1]] == Q_r.flatten()[x*Q_stride[0] + y*Q_stride[1]]
```
